```python
from transaction import Transaction, Mempool
 
 
def main() -> None:
    pool = Mempool()
 
    print("1) Phenyo -> Thato, amount = 10 ")
    phenyo_to_thato = Transaction(sender="Phenyo", recipient="Thato", amount=10.0)
    print("JSON payload:", phenyo_to_thato.to_json())
    accepted = pool.add(phenyo_to_thato)
    print(f"Accepted: {accepted}  |  Mempool size: {len(pool)}\n")
 
    print("2) Molete -> Thato, amount = -5 (invalid: negative amount)")
    molete_invalid = Transaction(sender="Molete", recipient="Thato", amount=-5.0)
    accepted = pool.add(molete_invalid)
    print(f"Accepted: {accepted}  |  Mempool size: {len(pool)}\n")
 
    print("3) Phenyo -> '' (invalid: empty recipient)")
    empty_recipient = Transaction(sender="Phenyo", recipient="", amount=2.0)
    accepted = pool.add(empty_recipient)
    print(f"Accepted: {accepted}  |  Mempool size: {len(pool)}\n")
 
    print("4) Round-trip check: serialise Phenyo's tx, then rebuild it ")
    payload = phenyo_to_thato.to_json()
    restored = Transaction.from_json(payload)
    print("Restored:", restored)
    assert restored.sender == "Phenyo"
    assert restored.recipient == "Thato"
    assert restored.amount == 10.0
    print("Round-trip preserved all fields correctly.\n")
 
    print("Final mempool contents (should contain only Phenyo -> Thato)")
    for tx in pool.get_all():
        print(" ", tx.to_dict())
 
    assert len(pool) == 1
    assert pool.get_all()[0].sender == "Phenyo"
    print("\nDemo complete: only the valid Phenyo -> Thato transaction is in the mempool.")
 
 
if __name__ == "__main__":
    main()
 
```
