```python
from transaction import Transaction, validate_transaction, Mempool
 
 
def test_schema_has_five_fields():
    tx = Transaction("Phenyo", "Thato", 10.0, timestamp=1_700_000_000.0, signature=None)
    keys = set(tx.to_dict().keys())
    assert keys == {"sender", "recipient", "amount", "timestamp", "signature"}
    print("test_schema_has_five_fields: PASS")
 
 
def test_json_round_trip_preserves_fields():
    tx = Transaction("Phenyo", "Thato", 10.0, timestamp=1_700_000_000.0, signature=None)
    restored = Transaction.from_json(tx.to_json())
    assert restored.sender == "Phenyo"
    assert restored.recipient == "Thato"
    assert restored.amount == 10.0
    assert restored.timestamp == 1_700_000_000.0
    assert restored.signature is None
    print("test_json_round_trip_preserves_fields: PASS")
 
 
def test_dict_round_trip():
    tx = Transaction("Phenyo", "Thato", 5.5, timestamp=42.0)
    rebuilt = Transaction.from_dict(tx.to_dict())
    assert rebuilt.to_dict() == tx.to_dict()
    print("test_dict_round_trip: PASS")
 
 
def test_valid_transaction_accepted():
    tx = Transaction("Phenyo", "Thato", 10.0, timestamp=1.0)
    assert validate_transaction(tx) is True
    print("test_valid_transaction_accepted: PASS")
 
 
def test_negative_amount_rejected():
    tx = Transaction("Molete", "Thato", -5.0, timestamp=1.0)
    assert validate_transaction(tx) is False
    print("test_negative_amount_rejected: PASS")
 
 
def test_zero_amount_rejected():
    tx = Transaction("Molete", "Thato", 0.0, timestamp=1.0)
    assert validate_transaction(tx) is False
    print("test_zero_amount_rejected: PASS")
 
 
def test_empty_sender_rejected():
    tx = Transaction("", "Thato", 1.0, timestamp=1.0)
    assert validate_transaction(tx) is False
    print("test_empty_sender_rejected: PASS")
 
 
def test_empty_recipient_rejected():
    tx = Transaction("Phenyo", "", 1.0, timestamp=1.0)
    assert validate_transaction(tx) is False
    print("test_empty_recipient_rejected: PASS")
 
 
def test_whitespace_only_recipient_rejected():
    tx = Transaction("Phenyo", "   ", 1.0, timestamp=1.0)
    assert validate_transaction(tx) is False
    print("test_whitespace_only_recipient_rejected: PASS")
 
 
def test_non_numeric_amount_rejected():
    tx = Transaction("Phenyo", "Thato", "ten", timestamp=1.0)  
    assert validate_transaction(tx) is False
    print("test_non_numeric_amount_rejected: PASS")
 
 
def test_bool_amount_rejected():
    # bool is a subclass of int in Python
    tx = Transaction("Phenyo", "Thato", True, timestamp=1.0)  
    assert validate_transaction(tx) is False
    print("test_bool_amount_rejected: PASS")
 
 
def test_mempool_accepts_valid_and_rejects_invalid():
    pool = Mempool()
    assert pool.add(Transaction("Phenyo", "Thato", 10.0, timestamp=1.0)) is True
    assert pool.add(Transaction("Molete", "Thato", -5.0, timestamp=1.0)) is False
    assert len(pool) == 1
    assert pool.get_all()[0].sender == "Phenyo"
    print("test_mempool_accepts_valid_and_rejects_invalid: PASS")
 
 
def test_mempool_get_all_returns_copy():
    pool = Mempool()
    pool.add(Transaction("Phenyo", "Thato", 10.0, timestamp=1.0))
    snapshot = pool.get_all()
    snapshot.append(Transaction("Evan", "Hayley", 999.0, timestamp=1.0))
    # Mutating the returned list must not affect the internal mempool
    assert len(pool) == 1
    print("test_mempool_get_all_returns_copy: PASS")
 
 
def run_all():
    tests = [
        test_schema_has_five_fields,
        test_json_round_trip_preserves_fields,
        test_dict_round_trip,
        test_valid_transaction_accepted,
        test_negative_amount_rejected,
        test_zero_amount_rejected,
        test_empty_sender_rejected,
        test_empty_recipient_rejected,
        test_whitespace_only_recipient_rejected,
        test_non_numeric_amount_rejected,
        test_bool_amount_rejected,
        test_mempool_accepts_valid_and_rejects_invalid,
        test_mempool_get_all_returns_copy,
    ]
    for t in tests:
        t()
    print(f"\nAll {len(tests)} tests passed.")
 
 
if __name__ == "__main__":
    run_all()
 
```
