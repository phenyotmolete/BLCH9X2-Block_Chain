```python
from Wallet_blchA4 import Wallet, canonical_tx_payload
```

```python
# Fixed Unix timestamp
TIMESTAMP = 1_700_000_000.0
AMOUNT = 10.0
TAMPERED_AMOUNT = 1000.0
```

```python
def main() -> None:
    phenyo = Wallet.create("Phenyo")
    thato = Wallet.create("Thato")

    print("*= Wallets =*")
    print(f"Phenyo address: {phenyo.address}")
    print(f"Thato address:  {thato.address}")
    print(f"Distinct addresses: {phenyo.address != thato.address}")
    print()
    # Step 2: canonical payload for Phenyo -> Thato
    payload = canonical_tx_payload(phenyo.address, thato.address, AMOUNT, TIMESTAMP)
    print("*= Canonical payload (Phenyo -> Thato, amount=10) =*")
    print(payload.decode("utf-8"))
    print
    # Step 3: Phenyo signs
    signature = phenyo.sign(payload)
    print("*= Signature =*")
    print(f"signature (hex, truncated): {signature[:32]}...")
    print()
# Step 4: verify original payload under Phenyo's key
    verify_original = Wallet.verify(payload, signature, phenyo.public_key_hex)

    # Step 5: alter/ mutate the amount, rebuild payload, verify again
    tampered_payload = canonical_tx_payload(phenyo.address, thato.address, TAMPERED_AMOUNT, TIMESTAMP)
    verify_tampered = Wallet.verify(tampered_payload, signature, phenyo.public_key_hex)

    # Step 6: verify the ORIGINAL payload under the wrong key (Thato's)
    verify_wrong_key = Wallet.verify(payload, signature, thato.public_key_hex)

    print("*= Verification checkpoints =*")
    print(f"verify(original payload, Phenyo's key):  {verify_original}")
    print(f"verify(tampered payload, Phenyo's key):  {verify_tampered}")
    print(f"verify(original payload, Thato's key):    {verify_wrong_key}")

    assert verify_original is True, "Original payload must verify under the signer's key"
    assert verify_tampered is False, "A mutated payload must fail verification"
    assert verify_wrong_key is False, "The wrong public key must fail verification"

    print()
    print("All checkpoints passed: signing binds Phenyo to these exact bytes,")
    print("and any change to the amount or the verifying key breaks that binding.")


if __name__ == "__main__":
    main()
```
